# Mobil Uygulama Abonelik Ekranı için A/B Testi: 
# Dönüşüm Oranını Artırma Analizi

# 1.Hipotezler

####  Null Hipotezi (HO)
Yeni abonelik ekranının dönüşüm oranı, mevcut ekrana kıyasla değişmez veya azalır.

#### Alternatif Hipotez (H1)
Yeni abonelik ekranı, mevcut ekrana kıyasla dönüşüm oranını artırır.

# 2. Dahil Edilecek Kullanıcılar

* Uygulamayı indiren ve abonelik ekranına ulaşan tüm kullanıcılar dahil edilir.
* Kullanıcılar rastgele iki gruba ayrılır:
1) Kontrol Grubu (A): Mevcut abonelik ekranını görenler
2) Deney Grubu (B): Yeni abonelik ekranını görenler

#### Kapsam:

Günlük 2000 yeni kullanıcı indiriyor, bunun %34’ü abonelik ekranına ulaşıyor → Günlük 680 kullanıcı teste dahil olabilir.

# 3.Hedef Metrik
#### Dönüşüm Oranı (Conversion Rate, CR): 
(Abonelik satın alan kullanıcılar) / (Abonelik teklifini gören kullanıcılar) 
* Abonelik ekranını gören kullanıcıların kaçının abonelik satın aldığını gösterir.
* Mevcut CR = %17, hedefimiz bunun artıp artmadığını görmek.

# Yardımcı Metrikler

#### Ortalama Gelir (ARPU - Average Revenue Per User): 
Toplam gelir / toplam kullanıcı 
* Bu metrik, indirimli abonelik seçeneğinin gelir üzerinde nasıl bir etki yarattığını anlamamızı sağlar.
#### Abonelik İptal Oranı (Churn Rate): 
* Kullanıcıların belirli bir süre sonra aboneliği iptal etme oranı
* İndirimli fiyatın kullanıcıların uzun vadeli abonelik süresi üzerindeki etkisini gösterir.
#### ROMI (Return on Marketing Investment): 
* Reklam harcamalarına göre getirinin değişimi 
* Yeni ekranın finansal sürdürülebilirliğini ölçmek için faydalı olabilir.


# 4. Örneklem Büyüklüğü ve Test Süresi

```python
import math
```

```python
# Verilen değerler
mevcut_donusum_orani = 0.17  
beklenen_donusum_orani = mevcut_donusum_orani * 1.20  # Beklenen dönüşüm oranı (%20 artış)
alpha = 0.05  # Anlamlılık seviyesi
power = 0.80  # İstatistiksel güç
haftalik_traffic = (2000*0.34)*7  # Günlük uygulama indiren 2000 kullanıcı * 0.34 ü abonelik ekrenına ulaşıyor. 
haftalik_traffic
```

```python
# İki örneklemli oran testi için gerekli örneklem büyüklüğünü hesaplamak:
# Oranların standart sapmalarını ve z-skorlarını belirlememiz gerekiyor.
# Anlamlılık seviyesi ve güç için z-skorları

# Alpha = 0.05 için z-skoru (iki kuyruklu)-anlamlılık seviyesinin standart normal dağılımdaki karşılığı.
z_alpha = 1.96 

# Beta = 0.20 için z-skoru - %80 güç için normal dağılımdaki karşılıktır. 
# Bu, istatistiksel testin doğru sonucu verme olasılığını ifade eder.
z_beta = 0.84  
```

```python
# Kontrol ve deney grubu için standart sapmalar

sd_control = math.sqrt(mevcut_donusum_orani * (1 - mevcut_donusum_orani))
sd_experiment = math.sqrt(beklenen_donusum_orani * (1 - beklenen_donusum_orani))
```

```python
# Örneklem büyüklüğü hesaplaması (iki örneklemli z-testi formülü)

orneklem_buyuklugu_grup_basina = (
    (z_alpha + z_beta) ** 2 * (sd_control ** 2 + sd_experiment ** 2)) / (beklenen_donusum_orani - mevcut_donusum_orani) ** 2
```

```python
# Toplam örneklem büyüklüğünü : (kontrol ve deney grubu)

toplam_orneklem_buyuklugu = 2 * orneklem_buyuklugu_grup_basina
```

```python
# Haftalık trafikle, testin gerekli örneklem büyüklüğüne ulaşması için gereken süre:

test_suresi_hafta = toplam_orneklem_buyuklugu / haftalik_traffic  
```

```python
print(f"Kontrol Grubu Dönüşüm Oranı: {mevcut_donusum_orani:.2%}")
print(f"Beklenen Dönüşüm Oranı (%20 Artış): {beklenen_donusum_orani:.2%}")
print(f"Gerekli Örneklem Büyüklüğü (Her Grup İçin): {round(orneklem_buyuklugu_grup_basina)}")
print(f"Toplam Gerekli Örneklem Büyüklüğü: {round(toplam_orneklem_buyuklugu)}")
print(f"Tahmini Test Süresi (Hafta): {round(test_suresi_hafta)} hafta")
```

# Sonuç: 
* Eğer yeni ekran tasarımı ile büyük bir değişiklik bekliyorsa (%20 artış beklenmesi gibi), bu test mükemmel bir seçenek sunar.
Kısa sürede (1 hafta) karar alınabilir ve hızlıca yeni tasarım yayına alınabilir.

* Ancak eğer abonelik oranındaki küçük artışlar bile değerli ise, MDE’yi %10 veya %7 olarak ayarlayıp tekrar hesaplamak yapmamız mantıklı olabilir.Çünkü %20’lik artış olmazsa, test başarısız olabilir ve testin tekrarlanması gerekebilir. Ancak bu durumda çok daha fazla katılımcıya ihtiyaç duyulur.
